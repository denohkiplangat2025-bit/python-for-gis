TUPLES
Are immutable sequences, meaning that once a tuple is created , its elements cannot be changed.
Tuples are useful for storing fixed colllections of items
For example , a tuple can be used to store the cordinates of a geographic point(latitude, longitude)

```python
point = (35.6895, 139.6917) #Tuple representing a geographic point(latitude, longitude)
```

```python
point
```

You can access elements in a tuple using indexing

```python
latitude = point[0]
longitude = point[1]
print(f"Latitude: {latitude}, Longitude: {longitude}")
```

```python
cities = (("Knoxville", "New York"), ("Nashville", "Chicago"))
```

```python
cities[1]
```

```python
cities[1][1]
```

SETS
sets are unordered collections of unique elements.
sets are useful when you need to store a collection of items but want to eliminate duplicates.
for example, you might want to sore a set of unique geographic regions visited during a survey.

```python
regions = ["North America", "Europe", "Asia", "South America", "Asia"]
regions = set(regions) #Converting the list to a set to remove duplicates
```

```python
regions
```

you can add a new region to the set

```python
regions.add("Africa") #Adding a new region to the set
print("Updated regions:", regions)
```

since sets do not allow duplicates , adding an existing region will not change the set:

```python
regions.add("Europe") #Attempting to add a duplicate region
print("Updated regions after adding duplicate:", regions)
```

```python
type(regions)
```

```python
region_list = list(regions) #Converting the set back to a list
```

```python
region_list
```

```python
region_list.append("Asia")
```

```python
region_list
```

```python
set(region_list)#Removing duplicates by converting the list back to a set
```

DICTIONARIES
dictionaries are collections of key-value-pairs, where each key is unique.
dictionaries are extremely useful for storing data that is not associated with specific identifiers, such as attribute data for geographic features.

for exmple, you can use a dictionary to store attributes of a geospatial feature, such as a city.

```python
city_attributes = {
    "name": "Tokyo",
    "population": 13929286,
    "coordinates": (35.6895, 139.6917),

}#Dictionary storing attributes of a city
```

```python
city_attributes
```

```python
city_attributes["population"] = 14000000 #Updating the population value in the dictionary
```

```python
city_attributes
```

```python
city_attributes["coordinates"] = (30, -100) 
#You can only replace a tuple, not a single element.
```

```python
city_attributes
```

you can access the values associated with specific keys:

```python
city_attributes["name"] = "tokyo"
```

```python
city_attributes["name"] = city_attributes["name"].upper()
```
