DICTIONARIES(dict)
are collections of key_value pairs

```python
feature_attributes = {
    "name": "Mount Fuji",
    "height_meters": 3776,
    "type": "Stratovolcano",
    "location": [35.3606, 138.7274],
}
```

ESCAPE CHARACTERS
are used to insert characters that are illegal in a string.
for exapmle,you can use escape character \n to insert a new line in a string.

```python
print("Hello World\nThis is a Python script.")
```

Another common escape character is \t, which inserts a tab in a string.

```python
print("This is the first line.\n\tThis is the second line.It is intended .")
```

If you want to include a single quote in a string , you can wrap the string in double quotes .
Alternativelly, you can use the escape character \ to include a single quote in a string.

```python
print("I say: \"hello world\"")
```

Converting latitude from degrees to radyians 
(required for some geospatial caluculations)

```python
import math
latitude = 35.6895
latitude_radians = math.radians(latitude)
print("Latitude in radians:", latitude_radians)
```
